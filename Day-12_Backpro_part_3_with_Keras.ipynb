```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns=['cgpa', 'profile_score', 'lpa'])
```

```python
df
```

```python
import tensorflow 
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input
```

```python
model = Sequential()

model.add(Input(shape=(2,)))
model.add(Dense(2,activation='linear'))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32), np.array([0.,0.],dtype=np.float32), np.array([[0.1],[0.1]], dtype=np.float32), np.array([0.],dtype=np.float32)]
```

```python
new_weights
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimiser = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error',optimizer=optimiser)
```

```python
model.fit(df.iloc[:,0:-1].values, df['lpa'].values, epochs=75, verbose=1, batch_size=1)
```
