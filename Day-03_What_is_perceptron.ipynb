# What is Perceptron

Perceptron হলো একটি supervised learning algorithm এবং একই সঙ্গে একটি simple artificial neuron model।
এটি training data থেকে weight ও bias শিখে input-কে সাধারণত দুইটি class-এর একটিতে classify করতে পারে।

তোমার কথার তিনটি অংশ দেখি:

- “একটি algorithm” — হ্যাঁ, Perceptron training algorithm ব্যবহার করে weight ও bias শেখে।
- “Deep Learning-এর fundamental unit” — সাধারণভাবে artificial neuron-কে neural network-এর basic computational unit বলা হয়। Perceptron হলো সেই neuron-এর একটি প্রাথমিক model।
- “Mathematical model” — হ্যাঁ, Perceptron-কে mathematical model হিসেবেও বর্ণনা করা যায়।

Note: একটি ছোট correction: Perceptron মূলত binary classification-এর জন্য তৈরি হয়েছিল। আর সব artificial neuron-ই Perceptron নয়।

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("files/placement.csv")
```

```python
data.shape
```

```python
data.head()
```

```python
sns.scatterplot(data=data, x=data['cgpa'], y=data['resume_score'], hue=data['placed'])
```

```python
X = data.drop('placed',axis=1)
y = data['placed']
```

```python
y
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
p.fit(X,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X.values, y.values, clf=p, legend=2)
```
