# What is backpropagation?
- Backpropagation is an algorithm used to train neural networks by calculating the gradients of the loss function with respect to the weights and biases.

সহজ বাংলায়:

- Backpropagation হলো এমন একটি algorithm, যা Loss Function-এর gradient calculate করে। এই gradient আমাদের জানায়, Loss কমানোর জন্য Weight ও Bias কীভাবে পরিবর্তন করতে হবে।

মনে রাখবে:
- Backpropagation: Gradient calculate করে।
- Gradient Descent: সেই gradient ব্যবহার করে Weight ও Bias update করে।

এই দুটি একসঙ্গে কাজ করে Neural Network-কে train করতে সাহায্য করে।

# How backpropagation works

- weight and bias ber korar jonno eita akta series of steps follow kore



- step0: initialize weight and bias 

       - oneke weight and bias randomly choose kore. abar onke w = 1 and b = 0 diya start kore

- step1: akta row nei, 1 no. point hisebe
- step2: predict kori oi student er lpa koto hobe; forward propagation[dot product]
        
       - suppose prediction asche "18 LPA" but actual hocche "3 LPA". tar but prediction wrong jeita w = 1 and b= 0 korar karone asche. akon eita update korte hobe loss function diya
- step3: choose kori loss function(MSE)
    
       - jahatu eita regression problem tai : MSE use korbo loss function hisebe
       - loss = (y-y')^2 = (3-18)^2 = 225; so eita amar error first student er jonno. akon ei error use kore  amar neural network ke bujate hobe je se bhul korse and kotota bhul korse taw se jane 225. akon amar weight and bais ke adjust korte hobe based on this error
       - ei loss = 225 ta hocche amar error. to ei error ta komanor jonno amar loss = (y-y')^2 er y' er moddhe changes ante hobe. tahole amar error or loss ta ke adjust korte parbo. and eita/y' komiya ba bariya adjust korte hobe error ke. so ei khetre dekhte passi ager prediction cilo 18. to eita komate hobe. 
       - The Question is: ei y' ke komano/barano kiser opor depend kore?
       - ans is: eita depend kore hidden layer er opor jei hocche y' = O21 = w11^2 * O11 + w21^2 * O12 + b21
       - orthat amar loss ke minimize korar jonno Lmin -> y^ ke change korte hobe -> ei y^ ei O21 -> ei O21 = w11^2 * O11 + w21^2 * O12 + b21 jeita 5ta jinis er opor depend kore. 
       - w11^2, o11, w21^2, o12  and b21 er opor . abar eikhane w11, w21, and bias b21 ke change kora jabe but O12 , and O11 ke direct korte parbo na . eita depend korbe b11, w11^1, w21^1 er opor. 

- step4: update weight and bias using gradient desecnt



       - ei weight and bias ke update korar jonno gradient desent er rule hocche
