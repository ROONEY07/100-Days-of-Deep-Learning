# Example below

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
or_data['input_1'] = [1,1,0,0]
or_data['input_2'] = [1,0,1,0]
or_data['Output'] = [1,1,1,0]
```

```python
and_data['input_1'] = [1,1,0,0]
and_data['input_2'] = [1,0,1,0]
and_data['Output'] = [1,0,0,0]
```

```python
xor_data['input_1'] = [1,1,0,0]
xor_data['input_2'] = [1,0,1,0]
xor_data['Output'] = [0,1,1,0]
```

```python
and_data
```

```python
sns.scatterplot(data=and_data, x=and_data['input_1'], y=and_data['input_2'], hue=and_data['Output'], s=200)
```

```python
or_data
```

```python
sns.scatterplot(data=or_data, x=or_data['input_1'], y=or_data['input_2'], hue=or_data['Output'], s=200)
```

```python
xor_data
```

```python
sns.scatterplot(data=xor_data, x=xor_data['input_1'], y=xor_data['input_2'], hue=xor_data['Output'], s=200)
```

```python
from sklearn.linear_model import Perceptron
clf1 = Perceptron()
clf2 = Perceptron()
clf3 = Perceptron()
```

```python
clf1.fit(and_data.iloc[:,0:2].to_numpy(dtype=int),and_data.iloc[:,-1].to_numpy(dtype=int))
clf2.fit(or_data.iloc[:,0:2].to_numpy(dtype=int),or_data.iloc[:,-1].to_numpy(dtype=int))
clf3.fit(xor_data.iloc[:,0:2].to_numpy(dtype=int),xor_data.iloc[:,-1].to_numpy(dtype=int))
```

```python
clf1.coef_
```

```python
clf1.intercept_
```

```python
x = np.linspace(-1,1,5)
x
```

```python
y = -x+1
```

```python
plt.plot(x,y)
sns.scatterplot(data=and_data, x=and_data['input_1'], y=and_data['input_2'], hue=and_data['Output'], s=200)
```

```python
x1 = np.linspace(-1,1,5)
y1 = -x+0.5
```

```python
plt.plot(x1,y1)
sns.scatterplot(data=or_data, x=or_data['input_1'], y=or_data['input_2'], hue=or_data['Output'], s=200)
```

```python
clf3.coef_
```

```python
clf3.intercept_
```

```python
x2 = np.linspace(-1,1,5)
y2 = -x+0.5
```

```python
plt.plot(x2,y2)
sns.scatterplot(data=xor_data, x=xor_data['input_1'], y=xor_data['input_2'], hue=xor_data['Output'], s=200)
```

```python
plot_decision_regions(xor_data.iloc[:,0:2].values, xor_data.iloc[:,-1].values, clf=clf3, legend=2)
```
