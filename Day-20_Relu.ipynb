# Dying Relu problem



- kisu kisu nuron er outout 0 hoya jai orthat weight update hoyna. means eita dead nueron bole, eita jodi 50% er beshi hoy, tahole amar jei dataset er opor kaj kortesi ta valo kore represent korte parbo na

0 hole , এই Neuron-এর Weight Gradient-এর মাধ্যমে Update হতে পারে না। এভাবে Neuron দীর্ঘ সময় নিষ্ক্রিয় থাকতে পারে। এর ফলে Network-এর Learning Capacity কমে যেতে পারে এবং Dataset-এর বিভিন্ন Pattern শেখার ক্ষমতা ক্ষতিগ্রস্ত হতে পারে।

# কেন Weight Update হয় না?

# when z1 is going to zero?

1. high learning rate
2. high negative bias

# How to solve it?

1. set low learning rate
2. set bias as positive like : 0.01
3. don't used relu. instead of relu used variance of relu
- variance of relu are two types of
  1. linear variance: leaky relu, parametric relu
  1. non-linear variance: Elu, Selu

# Leaky Relu
Leaky ReLU হলো ReLU Activation Function-এর একটি উন্নত সংস্করণ, যা মূলত Dying ReLU Problem কমানোর জন্য ব্যবহার করা হয়।






```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LeakyReLU, Input

model = Sequential([
    Input(shape=(10,)),

    Dense(128),
    LeakyReLU(negative_slope=0.01),

    Dense(64),
    LeakyReLU(negative_slope=0.01),

    Dense(1, activation='sigmoid')
])
```
