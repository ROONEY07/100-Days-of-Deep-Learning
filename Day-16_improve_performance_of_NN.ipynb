# How to improve Neural Networks?

1. Fine Tuning Neural Network hyperparameters
2. By solving problem:
   - Vanishing/Exploding problem
   - Not enough Data
   - slow traning
   - Overfitting

Neural Network-এ Hyperparameter হলো এমন কিছু configuration বা setting, যেগুলো আমরা Model Training শুরু করার আগে নিজেরা নির্ধারণ করি। এগুলো নির্ধারণ করে Neural Network কীভাবে শিখবে (Learn করবে)।

সহজ ভাষায়: Hyperparameter = Model Training কীভাবে হবে, তা নিয়ন্ত্রণ করার জন্য আগে থেকে নির্ধারিত সেটিংস।
