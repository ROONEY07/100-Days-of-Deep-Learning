# Types of neural networks
1. ***Feedforward Neural Network (FNN)***
- Data একদিকে Input থেকে Hidden Layer হয়ে Output Layer-এ যায়। সাধারণ classification ও regression problem-এ ব্যবহৃত হয়।

2. ***Convolutional Neural Network (CNN)***
- Image থেকে features, যেমন edge, shape ও pattern শিখতে পারে। Computer Vision-এ বহুল ব্যবহৃত।

3. ***Recurrent Neural Network (RNN)***
- Sequential data process করে এবং আগের ধাপের information hidden state-এর মাধ্যমে পরবর্তী ধাপে ব্যবহার করতে পারে।


4. ***LSTM (Long Short-Term Memory)***
- এটি RNN-এর একটি বিশেষ ধরন। Gate ব্যবহার করে দীর্ঘ sequence-এর গুরুত্বপূর্ণ information ধরে রাখতে সাহায্য করে।
- উদাহরণ: Speech ও time-series modeling।

Note: RNN and LSTM NLP based application e valo use khoy


5. ***Transformer Network***
- Self-attention ব্যবহার করে input-এর বিভিন্ন অংশের সম্পর্ক শেখে। আধুনিক NLP ও generative AI-তে গুরুত্বপূর্ণ।
- উদাহরণ: BERT, GPT, machine translation।

6. ***Generative Adversarial Network (GAN)***
- Generator ও Discriminator-এর প্রতিযোগিতামূলক training-এর মাধ্যমে synthetic data তৈরি করতে পারে।
- উদাহরণ: Synthetic image generation।

7. ***Autoencoder***
- Input-কে compact representation-এ encode করে এবং সেখান থেকে reconstruct করতে শেখে।
- উদাহরণ: Dimensionality reduction, anomaly detection।

Note: Autoenocer basically used to compress the file without loosing the quality. Eikhane input and output layer same node thake, but hidden layer e kom node thake, jate same kaj kora jai

8. ***Graph Neural Network (GNN)***
- Node ও edge দিয়ে গঠিত graph-structured data থেকে information শেখে।
- উদাহরণ: Social networks, molecular property prediction।

9. ***U-Net***
- Encoder-decoder structure ও skip connections ব্যবহার করে, বিশেষ করে image segmentation-এ পরিচিত।
- উদাহরণ: Medical image-এ tumor segmentation।

10. ***Radial Basis Function Network (RBFN)***
- Hidden layer-এ radial basis function ব্যবহার করে। Function approximation ও কিছু classification problem-এ ব্যবহৃত হয়।
