# Role of Loss function in Deep Learning

Loss Function in Deep Learning — Process
1. Initialize: Randomly Weight (\(W\)) এবং Bias (\(b\)) initialize করি।
2. Forward Propagation: Input features (CGPA, IQ) থেকে predicted output (\(\hat y\)) বের করি।
3. Calculate Loss: Actual output (\(y\)) ও predicted output (\(\hat y\))-এর পার্থক্য দিয়ে Loss calculate করি।
4. Backpropagation: Loss-এর gradient calculate করি, যা Weight ও Bias কীভাবে পরিবর্তন করতে হবে তা নির্দেশ করে।
5. Gradient Descent: Gradient ব্যবহার করে Weight ও Bias update করি।
6. Repeat: SGD ব্যবহার করলে updated Weight ও Bias দিয়ে পরবর্তী row-এর prediction করি। পুরো dataset process হলে একটি epoch সম্পন্ন হয়। Loss কমানো পর্যন্ত একাধিক epoch চালাই।
