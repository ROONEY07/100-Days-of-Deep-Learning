# What is Memoization?

Memoization হলো একটি programming technique, যেখানে কোনো function-এর একই input-এর জন্য আগেরবার calculate করা result সংরক্ষণ করে রাখা হয়, যাতে পরবর্তীতে একই input এলে আবার calculation করতে না হয়। Memoization মূলত computation time কমানোর technique। এটি Dynamic Programming, Recursive Algorithms এবং বিভিন্ন optimization problem-এ ব্যবহৃত হয়।

মনে রাখবে:
- Memoization → Store previous results.
- Reuse → Avoid repeated calculations.
- Benefit → Reduce computation time.

সহজ ভাষায়: Calculate once, store the result, reuse it later.

```python
import time
def fib(n):
    if n == 1 or n==0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
start = time.time()
fib(38)
print(time.time() - start)
```

```python
import time
def fib(n,d):
    if n in d:
        return d[n]
    else:
        d[n] = fib(n-1,d) + fib(n-2,d)
        return d[n]
```

```python
start = time.time()
d = {0:1, 1:1}
fib(100,d)
print(time.time() - start)
```
