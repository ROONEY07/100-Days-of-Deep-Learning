```python
def initialize_parameters(layer_dims):
    
    np.random.seed(3)
    parameter = {}
    L = len(layer_dims)
    
    for l in range(1,L):
        parameter['w' + str(l)] = np.ones((layer_dims[l-1],layer_dims[l]))*0.1
        parameter['b' + str(l)] = np.zeros((layer_dims[l],1))
    
    return parameter
```

```python
initialize_parameters([2,2,1])
```

```python
# calculate the output of given neuron
def linear_forward(A_prev,w,b):
    z = np.dot(w.T, A_prev) + b
    
    return z
```

# forward propagation

```python
def L_layer_forward(X,parameters):
    
    A = X
    L = len(parameters) // 2
    
    for l in range(1,L+1):
        
        A_prev = A
        w1 = parameters['w' + str(l)]
        b1 = parameters['b' + str(l)]
        
        print("A"+str(l-1)+": ", A_prev)
        print("W"+str(l)+": ", w1)
        print("b"+str(l)+": ", b1)
        print("--"*20)
        
        A = linear_forward(A_prev , w1, b1)
        print("A"+str(l)+": ",A)
        print("**"*20)
        
    return A, A_prev
```

```python
X = df[['cgpa', 'profile score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['lpa']].values[0][0]

# Parameter initialization
parameters = initialize_parameters([2,2,1])

# y_hat,A1 = L_layer_forward(X, parameters)
L_layer_forward(X, parameters)
```
