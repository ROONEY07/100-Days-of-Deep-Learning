Deep Learning-এর ভাষায় ANN-এর diagram-এর প্রতিটি node-কে সরাসরি Perceptron বলা ঠিক হবে না। চলো সহজভাবে বুঝি।

Node / Artificial Neuron: ANN-এর diagram-এ প্রতিটি circle-কে সাধারণত এভাবে বলা হয়।

Perceptron: Artificial neuron-এর একটি নির্দিষ্ট model, যা input নিয়ে weighted sum ও threshold/activation-এর মাধ্যমে output দেয়।

উদাহরণ:

ধরো, ANN-এর diagram-এ ১০টি circle আছে। সাধারণভাবে আমরা বলব, এখানে ১০টি node বা artificial neuron আছে। কিন্তু সেগুলো Perceptron model ব্যবহার করে তৈরি কি না, সেটা architecture-এর ওপর নির্ভর করে।

মনে রাখবে:

Every perceptron is an artificial neuron model, but not every artificial neuron is a perceptron.

# Deep Learning vs Machine Learning
1. Data Dependency: Eikhane DL er jonno large amount er Data proyojon
2. Hardware Dependency: DL e internal procces higly matrix calculation hoy, jar jonno GPU lage, insead of ML jeikhane normal CPU teo cole jai
3. Traning Time: Compare to ML , DL training time is High
4. Feature Selection: DL e Representation learning e kaj kore, jeikhane DL automatically feature Extract kore
5. Interpritibity: input dewar por output er result ta amar kon feature er jonno asche ta buja jai na. interpribility thake

# Deep Learning Libraries

1. **TensorFlow — developed by Google** : mostly used by indrustry level
2. **PyTorch — developed by Facebook (now Meta)** : mostly used by reaserch level

Google Brain team TensorFlow তৈরি করে এবং 2015 সালে open-source হিসেবে release করে। এটি Machine Learning ও Deep Learning model তৈরি, train এবং deploy করার কাজে ব্যবহৃত হয়।

প্রথমদিকে TensorFlow ব্যবহার করতে তুলনামূলকভাবে বেশি code লিখতে হতো এবং এর programming interface কিছু ব্যবহারকারীর কাছে জটিল মনে হতো।

এই সমস্যা সহজ করার জন্য **Keras** তৈরি করা হয়। Keras একটি high-level Deep Learning API, যা সহজে Neural Network তৈরি ও train করতে সাহায্য করে। এটি TensorFlow-এর সঙ্গে ব্যবহার করা যায়।

2017 সালে Google TensorFlow-এর জন্য Keras support চালু করে। এরপর **TensorFlow 2.0, 2019 সালে release হওয়ার সময় Keras-কে এর সঙ্গে আরও ঘনিষ্ঠভাবে integrate করা হয়**। TensorFlow 2.0-তে `tf.keras` ব্যবহার করে সহজে Deep Learning model তৈরি ও train করা যায়।

অন্যদিকে, PyTorch 2016 সালে open-source হিসেবে release হয়। এটি Facebook AI Research (FAIR) তৈরি করেছিল এবং flexible ও Python-friendly Deep Learning development-এর জন্য জনপ্রিয় হয়ে ওঠে।

- Azure AI Custom Vision helps create custom computer vision models.
- Apple Create ML helps train machine learning models for Apple applications.

# Deep Learning Architecture
- Deep Learning Architecture কী?
Neural network-এর layer, neuron এবং তাদের সংযোগ কীভাবে সাজানো হবে, সেই structural design-কে Neural Network Architecture বলে।
যেমন, তুমি যদি image classification-এর জন্য neural network বানাও, তাহলে কতগুলো layer থাকবে, কোন layer কীভাবে কাজ করবে—এসব নির্ধারণ করতে হবে। বিভিন্ন architecture নিয়ে experiment করে দেখতে হয় কোনটি তোমার problem ও dataset-এর জন্য ভালো কাজ করে।
কিন্তু প্রতিবার শুরু থেকে architecture design ও training করার প্রয়োজন নেই। আগে থেকে তৈরি এবং train করা model ব্যবহার করা যায়। eitake Pre-trained Model ও Transfer Learning bole
